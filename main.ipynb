```python
import pandas as pd
import numpy as np
import matplotlib as plt

from scipy.stats import kruskal

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.compose import make_column_selector

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import StackingRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from sklearn.model_selection import train_test_split
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
X_train = pd.read_csv('train.csv')
X_test = pd.read_csv('test.csv')
```

```python
X_train.head()
```

```python
X_train.info()
```

```python
X_train.shape
```

```python
y_train = X_train['SalePrice']
X_train = X_train.drop('SalePrice', axis=1)



X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

```

```python
X_train.isna().sum()[X_train.isna().sum() > 0].reset_index()
```

```python
cols_with_na =X_train.loc[:, X_train.isna().sum() > 600]
X_train = X_train.drop(cols_with_na, axis=1)
X_test = X_test.drop(cols_with_na, axis=1)
```

```python
pipe1 = Pipeline(steps=[('imputer', SimpleImputer(strategy='most_frequent')),
                      ('onehot', OneHotEncoder(handle_unknown='ignore'))])
```

```python
pipe2 = Pipeline(steps=[('numerical', SimpleImputer(strategy='mean')),
                      ('scaler', StandardScaler())])
```

```python

preprocessor = ColumnTransformer(
    [
    ('numerical', pipe2, make_column_selector(dtype_include='number')),
    ('categorical', pipe1, make_column_selector(dtype_include='object'))
    
]
)


pipeline = Pipeline([
    ('preprocessor', preprocessor)
])



X_train_transformed = pipeline.fit_transform(X_train)
X_val_transformed =pipeline.transform(X_val)
X_test_transformed = pipeline.transform(X_test)


transformed_columns = preprocessor.get_feature_names_out()

X_train_transformed = pd.DataFrame(X_train_transformed.toarray(), columns=transformed_columns)
X_val_transformed = pd.DataFrame(X_val_transformed.toarray(), columns=transformed_columns)
X_test_transformed = pd.DataFrame(X_test_transformed.toarray(), columns=transformed_columns)

X_train_transformed.head()
```

```python
X_train_transformed.shape
```

```python
X_train_transformed.info()
```

```python
base_models = [
    ('rf', RandomForestRegressor(n_estimators=100, random_state=42)),
    ('gbr', GradientBoostingRegressor(n_estimators=100, random_state=42)),
    ('svr', SVR(kernel='rbf', C=10))
]
```

```python
meta_model = Ridge(alpha=1.0)
```

```python
stacking_regressor = StackingRegressor(estimators=base_models, final_estimator=meta_model)
```

```python
stacking_regressor.fit(X_train_transformed, y_train)
y_pred = stacking_regressor.predict(X_val_transformed)

mse = mean_squared_error(y_val, y_pred)

print("Model MSE:", mse)

mae = mean_absolute_error(y_val, y_pred)

print("Model MAE:", mae)

rmse = np.sqrt(mse)

print ("Model RMSE:", rmse)


r2 = r2_score(y_val, y_pred)

print ("Model R2:", r2)
```

```python
baseline_pred = y_train.mean()
baseline_mse = np.mean((y_val - baseline_pred) ** 2)
print("Baseline MSE:", baseline_mse)
```
